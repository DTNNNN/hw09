```python
# HW09 兩場活動報名整理器 - 測試一（原始資料）

# 1. 原始資料設定（保留 board_raw 與 movie_raw 原值，不直接覆蓋）
activity_names = ("桌遊體驗", "電影欣賞")      # tuple：保存固定的兩個活動名稱
board_raw = " Amy, Ben, amy, Cora "            # str：桌遊原始報名文字
movie_raw = " ben, Dora, ERIC, ben "           # str：電影原始報名文字
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}  # dict：以活動名稱查詢每人單價

# 從 tuple 拆包讀取兩個活動名稱
act_board, act_movie = activity_names

# 2. 字串清理：先移除半形空白、統一轉小寫，再用逗號 split 成 list（保留重複與原順序）
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 3. 轉成 set 去除單場重複報名，並進行交集、聯集與雙向差集運算
board_set = set(board_list)                    # 桌遊去重集合
movie_set = set(movie_list)                    # 電影去重集合

both_set = board_set & movie_set               # 交集：兩場都參加
all_set = board_set | movie_set                # 聯集：至少參加一場
only_board_set = board_set - movie_set         # 差集：只參加桌遊
only_movie_set = movie_set - board_set         # 差集：只參加電影

# 4. 使用 sorted() 轉成依字母排序的 list，並用切片 [:2] 取得前兩位通知名單預覽
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

notify_preview = all_sorted[:2]                # 切片取得排序後的前兩位作為預覽

# 5. 使用 dict 保存摘要數據（金額在字典中保留數字型別，兩場依各自去重人數計費）
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie],
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 6. 使用 f-string 輸出完整摘要
print("HW09 活動報名摘要（測試一）")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{only_board_sorted}")
print(f"只參加電影者：{only_movie_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
